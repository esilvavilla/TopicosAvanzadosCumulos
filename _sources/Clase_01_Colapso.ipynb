## Perspectiva Histórica y Observaciones

Históricamente, el espacio entre las estrellas se consideraba vacío. Fue a mediados del siglo XX cuando Viktor Ambartsumian demostró que las asociaciones estelares OB se estaban expandiendo, lo que implicaba una formación reciente a partir de un medio gaseoso circundante.

El gran reto observacional de la formación estelar es que las nubes donde nacen las estrellas son extremadamente frías ({math}$T \sim 10 - 30$ K). A estas temperaturas, el principal constituyente, el hidrógeno molecular ({math}$H_2$), no emite radiación observable porque es una molécula simétrica sin momento dipolar eléctrico. Por lo tanto, dependemos de trazadores indirectos:

- Líneas Rotacionales del CO: El monóxido de carbono es la segunda molécula más abundante y posee un momento dipolar. Las colisiones con $H_2$ excitan el CO, que decae emitiendo fotones en el rango milimétrico (ej. la transición {math}$J=1 \rightarrow 0$ a {math}$\lambda \approx 2.6$ mm).

- Emisión del Polvo: El polvo interestelar (1% de la masa de la nube) absorbe luz ultravioleta y re-emite térmicamente en el infrarrojo lejano y submilimétrico. El flujo observado $S_\nu$ está dado por:
{math}$$S_\nu = \Omega_s B_\nu(T_d) (1 - e^{-\tau_\nu})$$
donde {math}$\Omega_s$ es el ángulo sólido, {math}$B_\nu(T_d)$ es la función de Planck para la temperatura del polvo {math}$T_d$, y {math}$\tau_\nu$ es la profundidad óptica.

## Estructura del Gas: Nubes, Núcleos y Grumos

Las GMCs no son esferas uniformes, sino que presentan una estructura fractal, modelada a menudo por turbulencia supersónica. La jerarquía se divide típicamente en:

- Nubes (GMCs): Tamaño {math}$\sim 10 - 100$ pc, masa {math}$\sim 10^4 - 10^6 M_\odot$, densidad media {math}$n \sim 10^2 - 10^3 \text{ cm}^{-3}$.

- Grumos (Clumps): Estructuras formadoras de cúmulos. Tamaño {math}$\sim 1$ pc, masa {math}$\sim 10^3 M_\odot$, densidad {math}$n \sim 10^4 \text{ cm}^{-3}$.

- Núcleos (Cores): Estructuras formadoras de estrellas individuales o sistemas múltiples. Tamaño {math}$\sim 0.1$ pc, masa {math}$\sim 0.1 - 10 M_\odot$, densidad {math}$n \gtrsim 10^5 \text{ cm}^{-3}$.

## Física del Equilibrio: Esferas Hidrostáticas Gravitantes

Para entender el inicio del colapso, primero debemos entender el equilibrio.

Consideremos una nube esférica, auto-gravitante e isoterma.
La ecuación de equilibrio hidrostático es:


{math}$$\frac{1}{\rho} \frac{dP}{dr} = -\frac{GM(r)}{r^2}$$

Asumiendo un gas ideal isotérmico, la presión térmica es:


{math}$$P = \rho c_s^2$$


donde la velocidad del sonido isotérmica es {math}$c_s = \sqrt{\frac{k_B T}{\mu m_H}}$.

Sustituyendo {math}$P$ y multiplicando por {math}$r^2/\rho$, luego derivando respecto a $r$, utilizamos la ecuación de continuidad de masa ({math}$\frac{dM}{dr} = 4\pi r^2 \rho$) para obtener:


{math}$$\frac{1}{r^2} \frac{d}{dr} \left( \frac{r^2 c_s^2}{\rho} \frac{d\rho}{dr} \right) = -4\pi G \rho$$

Esta es una forma de la ecuación de Poisson. Para resolverla, introducimos variables adimensionales para la densidad y el radio:


{math}$$\rho = \rho_c e^{-\psi}$$

{math}$$\xi = r \left( \frac{4\pi G \rho_c}{c_s^2} \right)^{1/2}$$


donde {math}$\rho_c$ es la densidad central y {math}$\psi$ es el potencial gravitacional adimensional. Sustituyendo, obtenemos la Ecuación Isotérmica de Lane-Emden:


{math}$$\frac{1}{\xi^2} \frac{d}{d\xi} \left( \xi^2 \frac{d\psi}{d\xi} \right) = e^{-\psi}$$

Con las condiciones de frontera {math}$\psi(0) = 0$ y $\psi'(0) = 0$.
La solución a esta ecuación modela una Esfera de Bonnor-Ebert. W.B. Bonnor (1956) y R. Ebert (1955) demostraron que esta esfera tiene una masa máxima estable que puede ser soportada contra el colapso gravitacional por una presión externa {math}$P_{ext}$:


{math}$$M_{BE} \approx 1.18 \frac{c_s^4}{G^{3/2} P_{ext}^{1/2}}$$


Si la masa de un núcleo supera {math}$M_{BE}$, ninguna configuración hidrostática es estable, y la nube debe colapsar.

## Inestabilidad: El Teorema del Virial y la Masa de Jeans

Una forma más general de derivar el criterio de colapso es usando el Teorema del Virial. Para un sistema cerrado de partículas, la ecuación del virial es:


{math}$$\frac{1}{2} \frac{d^2 I}{dt^2} = 2K + \Omega$$


donde {math}$I$ es el momento de inercia, {math}$K$ es la energía cinética total (térmica en nuestro caso) y {math}$\Omega$ es la energía potencial gravitacional. Para que la nube esté en equilibrio inminente de colapso, {math}$\frac{d^2 I}{dt^2} = 0$.

Para una nube esférica de masa {math}$M$ y radio {math}$R$:

Energía Térmica: {math}$K = \frac{3}{2} N k_B T = \frac{3}{2} \left( \frac{M}{\mu m_H} \right) k_B T$

Energía Gravitacional: {math}$\Omega = -\frac{3}{5} \frac{GM^2}{R}$ (asumiendo densidad uniforme para simplificar, el factor varía levemente para otras distribuciones).

El colapso ocurre cuando la gravedad domina a la presión térmica: {math}$\vert{}\Omega\vert{} > 2K$.


{math}$$\frac{3}{5} \frac{GM^2}{R} > 3 \frac{M}{\mu m_H} k_B T$$

Utilizando la relación de volumen para expresar {math}$R$ en términos de la densidad {math}$\rho = \frac{M}{\frac{4}{3}\pi R^3}$, sustituimos {math}$R = \left(\frac{3M}{4\pi\rho}\right)^{1/3}$.
Aislando la masa {math}$M$, encontramos la Masa de Jeans ({math}$M_J$):


{math}$$M > M_J \equiv \left( \frac{5 k_B T}{G \mu m_H} \right)^{3/2} \left( \frac{3}{4\pi\rho} \right)^{1/2}$$

Análogamente, la escala espacial de esta inestabilidad es la Longitud de Jeans ({math}$\lambda_J$):


{math}$$\lambda_J = c_s \left( \frac{\pi}{G\rho} \right)^{1/2}$$


Cualquier perturbación mayor a $\lambda_J$ será gravitacionalmente inestable y colapsará.
